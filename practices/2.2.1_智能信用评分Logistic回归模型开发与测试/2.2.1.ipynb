```python
import pandas as pd

data = pd.read_csv('data/finance数据集.csv')
data
```

```python
print(data.head())
```

```python
x = data.drop(['Unnamed: 0','SeriousDlqin2yrs'],axis=1)
y = data['SeriousDlqin2yrs']
```

```python
x
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)

from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(x_train,y_train)
```

```python
import pickle
with open('model.pkl','wb') as file:
    pickle.dump(model,file)
```

```python
y_predict = model.predict(x_test)
b = pd.DataFrame(y_predict,columns=['预测结果'])
```

```python
print(b)
```

```python
a = (y_test == y_predict).mean()
```

```python
print(a)
```

```python
from sklearn.metrics import classification_report
report = classification_report(y_test,y_predict,zero_division=1)
print(report)
```

```python
with open('report.txt','w') as file:
    file.write(report)
```

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)
x_resampled,y_resampled = smote.fit_resample(x_train,y_train)
```

```python
len(x_resampled)
```

```python
model.fit(x_resampled,y_resampled)
```

```python
y_repredict = model.predict(x_test)
```

```python
pd.DataFrame(y_repredict, columns=["预测结果"]).to_csv("预测结果.txt",index=False)
```
