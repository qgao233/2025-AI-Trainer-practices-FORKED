# 任务1：统计住院天数超过7天的患者数量及其占比

```python
data['RiskLevel'] = np.where(data['DaysInHospital'] > 7,1,0)
```

```python
riskCounts = data['RiskLevel'].value_counts()
```

```python
bmiArea = np.linspace(0,28,4)
print(bmiArea)
```

```python
bmiArea = np.append(bmiArea,[np.inf])
print(bmiArea)
```

```python
bmi_bins = np.array([0, 18.5, 24, 28, np.inf])
print(bmi_bins)
```

```python
print(len(bmiArea))
```

```python
bmi_labels = ['偏瘦', '正常', '超重', '肥胖']

data['bmiRange'] = pd.cut(data['BMI'],bins=bmiArea,labels=bmi_labels,right=False)
data.tail()
```

```python
print(data.groupby('bmiRange')['RiskLevel'].apply(lambda x: (x==1).mean()))
```

```python
print(data['bmiRange'].value_counts())
```

```python
a = np.linspace(0,100,5)
b = np.array(['小孩','青年','中年','老年'])
data['phase'] = pd.cut(data['Age'],bins=a,labels=b,right=False)
rate = data.groupby('phase')['RiskLevel'].apply(lambda x: (x == 1).mean())
num = data['phase'].value_counts()
print(rate,num)
```
